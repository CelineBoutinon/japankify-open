```python
import pandas as pd
import glob
import os
```

```python
# 1. Locate the extracted CSV file(s)
data_dir = './dataset'
csv_files = glob.glob(f'{data_dir}/*.csv')

if not csv_files:
    print("No CSV files found in the directory.")
else:
    for csv_path in csv_files:
        # 2. Define the target Parquet filename
        parquet_path = csv_path.replace('.csv', '.parquet')
        
        # 3. Load the CSV into memory
        print(f"Reading: {csv_path}")
        df = pd.read_csv(csv_path)
        
        # 4. Export to Parquet using the PyArrow engine
        print(f"Converting to: {parquet_path}")
        df.to_parquet(parquet_path, engine='pyarrow', index=False)
        
        # Display the file size difference
        csv_size = os.path.getsize(csv_path) / (1024 * 1024)
        pq_size = os.path.getsize(parquet_path) / (1024 * 1024)
        print(f"✅ Done! Size reduced from {csv_size:.2f} MB to {pq_size:.2f} MB.")
```
